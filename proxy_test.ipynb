# Тестирование reverse proxy из ноутбука

Запуск/остановка прокси одной функцией (код в `proxy_runner.py`, сам прокси — `main.py`).
Прокси работает в фоновом потоке, поэтому ноутбук остаётся свободным.

После запуска откройте адрес прокси в браузере и работайте с сайтом как обычно.

```python
from proxy_runner import start_proxy, stop_proxy
```

## Запуск

`start_proxy(target_url, *, port=8000, host="127.0.0.1", extra_domains=None, debug=False)`

- **target_url** — целевой сайт (обязательный аргумент);
- **port** — порт локального прокси (на этой машине порт 8000 занят другим сервисом — берите другой);
- **extra_domains** — доп. домены CDN/SSO (доступны как `/__<домен>__/...`); `None` = значения по умолчанию из `main.py`, `[]` = отключить;
- **debug** — подробные логи.

Повторный вызов `start_proxy` сам останавливает предыдущий инстанс.

```python
from proxy_runner import start_proxy, stop_proxy
server = start_proxy("https://ru.wikipedia.org", port=8899, extra_domains=[])
```

## Остановка

```python
stop_proxy()
```
